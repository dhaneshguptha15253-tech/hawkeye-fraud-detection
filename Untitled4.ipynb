```python
import random

random.seed(42)   # makes the random data the same every time

def make_data():
    data = []

    # 200 normal transactions
    for i in range(200):
        data.append({"amount": random.randint(100, 5000),
                     "hour": random.randint(7, 22),
                     "distance": random.randint(0, 30),
                     "txns_last_hour": random.randint(0, 2),
                     "foreign": 0,
                     "is_fraud": 0})

    # 6 fraud transactions
    for i in range(6):
        data.append({"amount": random.randint(15000, 90000),
                     "hour": random.randint(0, 4),
                     "distance": random.randint(300, 3000),
                     "txns_last_hour": random.randint(4, 8),
                     "foreign": random.randint(0, 1),
                     "is_fraud": 1})

    random.shuffle(data)   # mix them up

    for i in range(len(data)):
        data[i]["id"] = "T" + str(i + 1).zfill(3)
    return data

data = make_data()
print("Total transactions:", len(data))
print("First 3 transactions:")
for t in data[:3]:
    print(t)
```

```python
def learn(data):
    amounts = [t["amount"] for t in data]

    # average amount
    mean = sum(amounts) / len(amounts)

    # standard deviation = how much amounts usually vary from the average
    variance = sum((a - mean) ** 2 for a in amounts) / len(amounts)
    std = variance ** 0.5

    return mean, std

mean, std = learn(data)
print("Average amount: Rs", round(mean))
print("Spread (std)  : Rs", round(std))
```

```python
def risk_score(t, mean, std):
    score = 0
    reasons = []

    z = (t["amount"] - mean) / std   # how far the amount is from normal

    if z > 3:
        score += 40
        reasons.append("amount far above normal")
    elif z > 2:
        score += 20
        reasons.append("amount above normal")

    if t["hour"] < 5:
        score += 20
        reasons.append("night-time transaction")

    if t["distance"] > 200:
        score += 20
        reasons.append("far from home location")

    if t["txns_last_hour"] >= 4:
        score += 20
        reasons.append("many transactions in 1 hour")

    if t["foreign"] == 1:
        score += 20
        reasons.append("foreign transaction")

    return score, reasons


THRESHOLD = 50   # score of 50 or more = suspicious

flagged = []
for t in data:
    s, r = risk_score(t, mean, std)
    t["score"] = s
    t["reasons"] = r
    if s >= THRESHOLD:
        flagged.append(t)

print("Flagged suspicious:", len(flagged))
for t in sorted(flagged, key=lambda x: x["score"], reverse=True):
    print(t["id"], "| Rs", t["amount"], "| score", t["score"], "|", ", ".join(t["reasons"]))
```

```python
while True:
    print("\n===== FRAUD DETECTION MENU =====")
    print("1. Summary and accuracy")
    print("2. Show flagged (suspicious) transactions")
    print("3. Check a new transaction")
    print("4. Exit")
    choice = input("Choose 1-4: ")

    if choice == "1":
        caught = sum(1 for t in flagged if t["is_fraud"] == 1)
        total_fraud = sum(1 for t in data if t["is_fraud"] == 1)
        false_alarms = len(flagged) - caught
        print("Total transactions :", len(data))
        print("Flagged suspicious :", len(flagged))
        print("Real frauds caught :", caught, "out of", total_fraud)
        print("False alarms       :", false_alarms)
        print("Amount at risk     : Rs", sum(t["amount"] for t in flagged))

    elif choice == "2":
        for t in sorted(flagged, key=lambda x: x["score"], reverse=True):
            print(t["id"], "| Rs", t["amount"], "| score", t["score"], "|", ", ".join(t["reasons"]))

    elif choice == "3":
        new = {"amount": int(input("Amount (Rs): ")),
               "hour": int(input("Hour of day (0-23): ")),
               "distance": int(input("Distance from home (km): ")),
               "txns_last_hour": int(input("Transactions in last hour: ")),
               "foreign": int(input("Foreign? (1 = yes, 0 = no): "))}
        s, r = risk_score(new, mean, std)
        if s >= THRESHOLD:
            print("SUSPICIOUS! Risk score", s, "->", ", ".join(r))
        else:
            print("Looks normal. Risk score", s)

    elif choice == "4":
        print("Bye!")
        break

    else:
        print("Please enter 1, 2, 3 or 4")
```
